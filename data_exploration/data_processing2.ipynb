```python

all_downloads_csv_path = r"C:/Users/Hychiro/Documents/Mestrado/Mestrado/Mestrado/Review/SLR-BCI-temporal-filters/structured_data/secondStepBaixados.csv"

path = r"C:/Users/Hychiro/Documents/Mestrado/Mestrado/Mestrado/Review/SLR-BCI-temporal-filters/structured_data/firstStepOutput.csv"


# =========================
# carregar datasets
# =========================

df_downloads = pd.read_csv(
    all_downloads_csv_path,
    encoding="utf-8"
)

df_first = pd.read_csv(
    path,
    encoding="utf-8"
)


# =========================
# normalizar títulos
# =========================

df_downloads["_titulo_normalizado"] = (
    df_downloads["Titulo"]
    .astype(str)
    .str.strip()
    .str.lower()
)

df_first["_titulo_normalizado"] = (
    df_first["Titulo"]
    .astype(str)
    .str.strip()
    .str.lower()
)


# =========================
# remover duplicatas
# mantendo apenas 1 cópia
# =========================

df_downloads = df_downloads.drop_duplicates(
    subset="_titulo_normalizado",
    keep="first"
)

df_first = df_first.drop_duplicates(
    subset="_titulo_normalizado",
    keep="first"
)


# =========================
# conjuntos de títulos
# =========================

titulos_downloads = set(df_downloads["_titulo_normalizado"])
titulos_first = set(df_first["_titulo_normalizado"])


# =========================
# artigos presentes nos dois
# =========================

titulos_em_ambos = titulos_downloads & titulos_first


# =========================
# somente no secondStepBaixados
# =========================

somente_downloads = titulos_downloads - titulos_first


# =========================
# somente no firstStepOutput
# =========================

somente_first = titulos_first - titulos_downloads


# =========================
# resultados
# =========================

print("Artigos no secondStepBaixados:", len(titulos_downloads))
print("Artigos no firstStepOutput:", len(titulos_first))
print("Artigos presentes nos dois:", len(titulos_em_ambos))
print("Somente no secondStepBaixados:", len(somente_downloads))
print("Somente no firstStepOutput:", len(somente_first))


print("\n--- SOMENTE NO secondStepBaixados ---")

for titulo in sorted(somente_downloads):
    print(titulo)


print("\n--- SOMENTE NO firstStepOutput ---")

for titulo in sorted(somente_first):
    print(titulo)

```
